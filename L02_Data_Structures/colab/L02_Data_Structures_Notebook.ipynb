## 1. Lists - Ordered Sequences

Lists store ordered collections of items. They are:
- **Ordered**: Items maintain their position
- **Indexed**: Access by position (0, 1, 2, ...)
- **Mutable**: Can be changed after creation

````
# Creating lists
prices = [150.0, 165.5, 148.25, 172.0, 169.75]
tickers = ["AAPL", "MSFT", "GOOGL", "AMZN", "SPY"]

print("Prices:", prices)
print("Tickers:", tickers)
print("Length:", len(prices))
````

````
# Indexing - accessing elements
print("First price (index 0):", prices[0])
print("Third price (index 2):", prices[2])
print("Last price (index -1):", prices[-1])
print("Second from end:", prices[-2])
````

## 2. Slicing - Extracting Portions

Syntax: `list[start:end:step]`
- `start`: First index (included)
- `end`: Last index (excluded)
- `step`: Skip interval (optional)

````
# Slicing examples
tickers = ["AAPL", "MSFT", "GOOGL", "AMZN", "SPY", "TSLA", "NVDA"]

print("tickers[0:3]:", tickers[0:3])   # First 3
print("tickers[2:5]:", tickers[2:5])   # Middle portion
print("tickers[::2]:", tickers[::2])   # Every 2nd element
print("tickers[-3:]:", tickers[-3:])   # Last 3
````

## 3. Dictionaries - Key-Value Pairs

Dictionaries store data as key-value pairs:
- **Keys**: Unique identifiers (strings, numbers)
- **Values**: Any data type
- **Fast lookup**: O(1) access time

````
# Creating dictionaries
portfolio = {
    "AAPL": 150.50,
    "MSFT": 340.00,
    "GOOGL": 125.75,
    "AMZN": 165.00
}

print("Portfolio:", portfolio)
print("AAPL price:", portfolio["AAPL"])
print("Keys:", list(portfolio.keys()))
print("Values:", list(portfolio.values()))
````

````
# Modifying dictionaries
portfolio["AAPL"] = 155.00      # Update value
portfolio["TSLA"] = 250.00      # Add new key

print("Updated portfolio:", portfolio)
print("Is MSFT in portfolio?", "MSFT" in portfolio)
````

## 4. List Comprehensions

Compact way to create lists:
```python
[expression for item in iterable if condition]
```

````
# List comprehension examples
prices = [150, 165, 148, 172, 169]

# Double all prices
doubled = [p * 2 for p in prices]
print("Doubled:", doubled)

# Filter prices > 155
high_prices = [p for p in prices if p > 155]
print("High prices:", high_prices)

# Calculate returns (simple)
returns = [(prices[i] - prices[i-1]) / prices[i-1] * 100 
           for i in range(1, len(prices))]
print("Returns (%):", [round(r, 2) for r in returns])
````

## Finance Application: Portfolio Tracker

Let's build a simple portfolio tracker combining everything we learned.

````
# Portfolio data
portfolio = {
    "AAPL": {"shares": 100, "buy_price": 145.00, "current_price": 150.50},
    "MSFT": {"shares": 50, "buy_price": 320.00, "current_price": 340.00},
    "GOOGL": {"shares": 75, "buy_price": 120.00, "current_price": 125.75}
}

# Calculate total value
total_value = sum(
    data["shares"] * data["current_price"] 
    for data in portfolio.values()
)
print(f"Total Portfolio Value: ${total_value:,.2f}")

# Calculate profit per stock
for ticker, data in portfolio.items():
    profit = data["shares"] * (data["current_price"] - data["buy_price"])
    print(f"{ticker}: ${profit:,.2f} profit")
````
