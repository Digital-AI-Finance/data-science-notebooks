## 1. Function Basics

Functions encapsulate reusable code:
- **def**: keyword to define a function
- **Parameters**: inputs to the function
- **return**: sends value back to caller

````python
# Simple function definition
def calculate_return(buy_price, sell_price):
    """Calculate percentage return from buy to sell price."""
    return (sell_price - buy_price) / buy_price * 100

# Calling the function
ret = calculate_return(100, 115)
print(f"Return: {ret}%")
````

````python
# Function with default parameters
def annualize_return(daily_return, trading_days=252):
    """Annualize a daily return."""
    return daily_return * np.sqrt(trading_days)

# Using default value
print(f"Annualized (default): {annualize_return(0.05):.2f}%")

# Overriding default
print(f"Annualized (260 days): {annualize_return(0.05, 260):.2f}%")
````

## 2. Multiple Return Values

Functions can return multiple values using tuples.

````python
# Returning multiple values
def calculate_stats(prices):
    """Calculate mean and standard deviation of prices."""
    mean = np.mean(prices)
    std = np.std(prices)
    return mean, std

prices = [150.0, 152.5, 148.0, 155.0, 153.5]
avg, volatility = calculate_stats(prices)
print(f"Mean: ${avg:.2f}, Volatility: ${volatility:.2f}")
````

## 3. Variable Scope

Variables have different scopes:
- **Local**: exists only inside the function
- **Global**: accessible everywhere

````python
# Global variable
TAX_RATE = 0.15

def calculate_tax(income):
    """Calculate tax using global rate."""
    # Can READ global variable
    tax = income * TAX_RATE
    return tax

print(f"Tax on $10,000: ${calculate_tax(10000):.2f}")
````

````python
# Local variables
def process_trade(price):
    result = price * 1.01  # Local variable
    commission = 0.005     # Local variable
    return result * (1 - commission)

# print(result)  # This would error - result doesn't exist here!
print(f"Processed: ${process_trade(100):.2f}")
````

## 4. Docstrings

Document your functions with docstrings for clarity.

````python
def sharpe_ratio(returns, risk_free_rate=0.02):
    """
    Calculate the Sharpe ratio for a series of returns.
    
    Parameters:
    -----------
    returns : list or array
        Daily or periodic returns
    risk_free_rate : float, optional
        Annual risk-free rate (default 2%)
    
    Returns:
    --------
    float
        Annualized Sharpe ratio
    """
    excess_returns = np.array(returns) - risk_free_rate/252
    return np.mean(excess_returns) / np.std(excess_returns) * np.sqrt(252)

# Access the docstring
help(sharpe_ratio)
````

## Finance Application: Functions Library

Build a reusable library of finance functions.

````python
# Finance functions library

def simple_return(start_price, end_price):
    """Calculate simple percentage return."""
    return (end_price - start_price) / start_price

def log_return(start_price, end_price):
    """Calculate logarithmic return."""
    return np.log(end_price / start_price)

def portfolio_return(weights, returns):
    """Calculate weighted portfolio return."""
    return np.sum(np.array(weights) * np.array(returns))

def max_drawdown(prices):
    """Calculate maximum drawdown from peak."""
    peak = prices[0]
    max_dd = 0
    for price in prices:
        if price > peak:
            peak = price
        dd = (peak - price) / peak
        if dd > max_dd:
            max_dd = dd
    return max_dd

# Test the library
print(f"Simple return: {simple_return(100, 110):.2%}")
print(f"Log return: {log_return(100, 110):.4f}")
print(f"Portfolio return: {portfolio_return([0.5, 0.3, 0.2], [0.05, 0.03, -0.02]):.2%}")
print(f"Max drawdown: {max_drawdown([100, 110, 95, 105, 90]):.2%}")
````
