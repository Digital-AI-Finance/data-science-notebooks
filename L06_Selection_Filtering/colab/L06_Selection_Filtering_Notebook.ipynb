````
import pandas as pd
import numpy as np
np.random.seed(42)

# Create sample stock data
dates = pd.date_range('2024-01-01', periods=20, freq='B')
df = pd.DataFrame({
    'Date': dates,
    'AAPL': 185 + np.random.randn(20).cumsum() * 2,
    'MSFT': 375 + np.random.randn(20).cumsum() * 3,
    'GOOGL': 140 + np.random.randn(20).cumsum() * 2,
    'Volume': np.random.randint(1000000, 5000000, 20)
})
df = df.set_index('Date')
print('Sample data created!')
df.head()
````

## 2. iloc vs loc

- **iloc**: Integer position (0, 1, 2...)
- **loc**: Label-based (dates, names)

````
# iloc - position based
print('First row:')
print(df.iloc[0])
print('\nFirst 3 rows, first 2 columns:')
print(df.iloc[0:3, 0:2])
````

````
# loc - label based
print('Row by date:')
print(df.loc['2024-01-01'])
print('\nSelect AAPL column:')
print(df.loc[:, 'AAPL'].head())
````

## 3. Boolean Filtering

Filter rows based on conditions.

````
# Simple condition
high_aapl = df[df['AAPL'] > 190]
print(f'Rows where AAPL > 190: {len(high_aapl)}')
print(high_aapl)
````

````
# Multiple conditions with & (AND)
filtered = df[(df['AAPL'] > 185) & (df['MSFT'] > 375)]
print(f'AAPL > 185 AND MSFT > 375: {len(filtered)} rows')
print(filtered.head())
````

````
# Multiple conditions with | (OR)
filtered = df[(df['AAPL'] > 195) | (df['MSFT'] > 385)]
print(f'AAPL > 195 OR MSFT > 385: {len(filtered)} rows')
````

More readable syntax for complex filters.

## Finance Application: Stock Screener

Build a simple stock screener.

````
# Stock screener
def screen_stocks(df, min_price=None, max_price=None, min_volume=None):
    result = df.copy()
    if min_price:
        result = result[result['AAPL'] >= min_price]
    if max_price:
        result = result[result['AAPL'] <= max_price]
    if min_volume:
        result = result[result['Volume'] >= min_volume]
    return result

# Screen for AAPL between 185-195 with high volume
screened = screen_stocks(df, min_price=185, max_price=195, min_volume=2000000)
print(f'Screened stocks: {len(screened)} days')
print(screened)
````
