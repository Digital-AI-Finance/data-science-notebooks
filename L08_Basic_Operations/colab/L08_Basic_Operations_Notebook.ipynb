## Learning Objectives
- Apply functions to columns and rows
- Sort and rank data
- Calculate returns and rolling statistics
- Use value_counts for categorical analysis

````
import pandas as pd
import numpy as np
np.random.seed(42)

# Create sample stock data
dates = pd.date_range('2024-01-01', periods=50, freq='B')
df = pd.DataFrame({
    'Date': dates,
    'AAPL': 185 + np.random.randn(50).cumsum() * 2,
    'MSFT': 375 + np.random.randn(50).cumsum() * 3,
    'Volume': np.random.randint(1000000, 5000000, 50),
    'Sector': np.random.choice(['Tech', 'Finance', 'Health'], 50)
})
df = df.set_index('Date')
print('Data ready!')
df.head()
````

## 1. Apply Function
Apply custom functions to columns or rows.

````
# Apply to single column
df['AAPL_rounded'] = df['AAPL'].apply(lambda x: round(x, 2))
print(df[['AAPL', 'AAPL_rounded']].head())
````

````
# Categorize prices
def categorize_price(price):
    if price > 190:
        return 'High'
    elif price > 180:
        return 'Medium'
    else:
        return 'Low'

df['AAPL_Category'] = df['AAPL'].apply(categorize_price)
print(df['AAPL_Category'].value_counts())
````

## 2. Sorting
Order data by values.

````
# Sort by single column
sorted_df = df.sort_values('AAPL', ascending=False)
print('Top 5 AAPL prices:')
print(sorted_df[['AAPL']].head())
````

````
# Sort by multiple columns
sorted_df = df.sort_values(['Sector', 'AAPL'], ascending=[True, False])
print(sorted_df[['Sector', 'AAPL']].head(10))
````

## 3. Calculating Returns
Compute percentage changes.

````
# Simple returns
df['AAPL_Return'] = df['AAPL'].pct_change()
print(df[['AAPL', 'AAPL_Return']].head(10))
````

````
# Log returns
df['AAPL_LogReturn'] = np.log(df['AAPL'] / df['AAPL'].shift(1))
print('Mean daily return:', df['AAPL_Return'].mean())
print('Return volatility:', df['AAPL_Return'].std())
````

## 4. Rolling Windows
Calculate moving statistics.

````
# 5-day moving average
df['AAPL_MA5'] = df['AAPL'].rolling(window=5).mean()
df['AAPL_MA20'] = df['AAPL'].rolling(window=20).mean()
print(df[['AAPL', 'AAPL_MA5', 'AAPL_MA20']].tail(10))
````

````
# Rolling volatility
df['AAPL_Vol20'] = df['AAPL_Return'].rolling(window=20).std() * np.sqrt(252)
print('Annualized 20-day volatility:')
print(df['AAPL_Vol20'].tail())
````

## 5. Value Counts
Analyze categorical data.

````
# Count by sector
print(df['Sector'].value_counts())
print('\nPercentages:')
print(df['Sector'].value_counts(normalize=True))
````
