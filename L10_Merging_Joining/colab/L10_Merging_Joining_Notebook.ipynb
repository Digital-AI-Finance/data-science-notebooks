## Learning Objectives
- Use pd.concat() to stack DataFrames
- Apply pd.merge() for SQL-style joins
- Understand join types (inner, left, right, outer)
- Combine multiple data sources

````python
import pandas as pd
import numpy as np
np.random.seed(42)

# Create sample stock price data
prices = pd.DataFrame({
    'Symbol': ['AAPL', 'MSFT', 'GOOGL', 'AMZN'],
    'Price': [185.50, 375.20, 140.80, 178.25],
    'Volume': [50000000, 25000000, 30000000, 45000000]
})

# Create company info data
info = pd.DataFrame({
    'Symbol': ['AAPL', 'MSFT', 'GOOGL', 'META'],
    'Sector': ['Technology', 'Technology', 'Technology', 'Technology'],
    'MarketCap': [2850, 2780, 1750, 1200]  # Billions
})

print('Prices DataFrame:')
print(prices)
print('\nInfo DataFrame:')
print(info)
````

## 1. pd.concat() - Stacking DataFrames
Combine DataFrames vertically or horizontally.

````python
# Create two DataFrames to stack
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})

# Vertical concatenation (axis=0)
vertical = pd.concat([df1, df2], ignore_index=True)
print('Vertical concatenation:')
print(vertical)
````

````python
# Horizontal concatenation (axis=1)
df3 = pd.DataFrame({'C': [10, 20]})
horizontal = pd.concat([df1, df3], axis=1)
print('Horizontal concatenation:')
print(horizontal)
````

## 2. pd.merge() - SQL-Style Joins
Combine DataFrames on common columns.

````python
# Inner join (default) - only matching keys
inner = pd.merge(prices, info, on='Symbol')
print('Inner join (only matching symbols):')
print(inner)
````

````python
# Left join - all from left, matching from right
left = pd.merge(prices, info, on='Symbol', how='left')
print('Left join (all prices, matching info):')
print(left)
````

````python
# Outer join - all keys from both
outer = pd.merge(prices, info, on='Symbol', how='outer')
print('Outer join (all symbols from both):')
print(outer)
````

## 3. Merging on Different Column Names

````python
# Different column names
df_left = pd.DataFrame({'ticker': ['AAPL', 'MSFT'], 'price': [185, 375]})
df_right = pd.DataFrame({'symbol': ['AAPL', 'MSFT'], 'sector': ['Tech', 'Tech']})

merged = pd.merge(df_left, df_right, left_on='ticker', right_on='symbol')
print('Merge with different column names:')
print(merged)
````

## 4. Merging on Multiple Keys

````python
# Create data with date and symbol
daily_prices = pd.DataFrame({
    'Date': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    'Symbol': ['AAPL', 'MSFT', 'AAPL', 'MSFT'],
    'Price': [185, 375, 186, 376]
})

daily_volume = pd.DataFrame({
    'Date': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    'Symbol': ['AAPL', 'MSFT', 'AAPL', 'MSFT'],
    'Volume': [50e6, 25e6, 55e6, 28e6]
})

# Merge on multiple keys
merged = pd.merge(daily_prices, daily_volume, on=['Date', 'Symbol'])
print('Merge on multiple keys:')
print(merged)
````

## 5. df.join() - Index-Based Joining

````python
# Create indexed DataFrames
df1 = pd.DataFrame({'A': [1, 2, 3]}, index=['a', 'b', 'c'])
df2 = pd.DataFrame({'B': [4, 5, 6]}, index=['a', 'b', 'd'])

# Join on index
joined = df1.join(df2, how='outer')
print('Join on index:')
print(joined)
````
