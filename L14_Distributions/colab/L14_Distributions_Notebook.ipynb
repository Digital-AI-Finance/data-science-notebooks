## Learning Objectives
- Understand normal, t, and binomial distributions
- Work with PDF and CDF
- Create Q-Q plots for normality testing
- Apply distributions to financial data

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
np.random.seed(42)

# Generate sample stock returns
returns = np.random.randn(1000) * 0.02  # ~2% daily vol
print(f'Sample size: {len(returns)}')
print(f'Mean: {returns.mean():.4f}')
print(f'Std: {returns.std():.4f}')
````

## 1. Normal Distribution
The bell curve - most common distribution.

````python
# Create normal distribution
mu, sigma = 0, 1
x = np.linspace(-4, 4, 100)

# PDF and CDF
pdf = stats.norm.pdf(x, mu, sigma)
cdf = stats.norm.cdf(x, mu, sigma)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.plot(x, pdf)
ax1.fill_between(x, pdf, alpha=0.3)
ax1.set_title('PDF - Probability Density')
ax2.plot(x, cdf)
ax2.set_title('CDF - Cumulative Probability')
plt.tight_layout()
plt.show()
````

````python
# 68-95-99.7 Rule
print('Probability within:')
print(f'  1 std: {stats.norm.cdf(1) - stats.norm.cdf(-1):.1%}')
print(f'  2 std: {stats.norm.cdf(2) - stats.norm.cdf(-2):.1%}')
print(f'  3 std: {stats.norm.cdf(3) - stats.norm.cdf(-3):.1%}')
````

## 2. t-Distribution
Heavier tails than normal - better for financial data.

````python
# Compare normal vs t-distribution
x = np.linspace(-5, 5, 100)
normal = stats.norm.pdf(x)
t_5 = stats.t.pdf(x, df=5)
t_30 = stats.t.pdf(x, df=30)

plt.figure(figsize=(10, 5))
plt.plot(x, normal, label='Normal')
plt.plot(x, t_5, label='t (df=5)', linestyle='--')
plt.plot(x, t_30, label='t (df=30)', linestyle=':')
plt.legend()
plt.title('Normal vs t-Distribution')
plt.show()
````

## 3. Q-Q Plots
Check if data follows a theoretical distribution.

````python
# Q-Q plot for our returns
fig, ax = plt.subplots(figsize=(8, 6))
stats.probplot(returns, dist='norm', plot=ax)
ax.set_title('Q-Q Plot: Returns vs Normal')
plt.show()
````

## 4. Goodness of Fit

````python
# Jarque-Bera test for normality
jb_stat, jb_pvalue = stats.jarque_bera(returns)
print(f'Jarque-Bera statistic: {jb_stat:.4f}')
print(f'p-value: {jb_pvalue:.4f}')
print(f'Normal? {"Yes" if jb_pvalue > 0.05 else "No"}')
````
