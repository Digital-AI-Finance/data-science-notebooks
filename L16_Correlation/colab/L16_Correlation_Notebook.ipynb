## Learning Objectives
- Calculate Pearson and Spearman correlation
- Create correlation heatmaps
- Understand correlation vs causation
- Apply to portfolio diversification

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
np.random.seed(42)

# Generate correlated stock returns
n = 252  # Trading days
market = np.random.randn(n) * 0.01
stock_a = 0.8 * market + 0.2 * np.random.randn(n) * 0.01
stock_b = 0.5 * market + 0.5 * np.random.randn(n) * 0.01
stock_c = -0.3 * market + 0.7 * np.random.randn(n) * 0.01

df = pd.DataFrame({'Market': market, 'Stock_A': stock_a, 
                   'Stock_B': stock_b, 'Stock_C': stock_c})
print('Data ready!')
df.head()
````

## 1. Pearson Correlation
Measures linear relationship.

````python
# Correlation between two columns
corr_ab = df['Stock_A'].corr(df['Stock_B'])
print(f'Correlation A-B: {corr_ab:.4f}')

# Full correlation matrix
print('\nCorrelation Matrix:')
print(df.corr().round(3))
````

````python
# Scatter plot with correlation
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(df['Market'], df['Stock_A'], alpha=0.5)
ax.set_xlabel('Market Return')
ax.set_ylabel('Stock A Return')
ax.set_title(f'Market vs Stock A (r = {df["Market"].corr(df["Stock_A"]):.3f})')
plt.show()
````

## 2. Correlation Heatmap

````python
# Heatmap of correlations
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(), annot=True, cmap='RdBu_r', center=0,
            vmin=-1, vmax=1, fmt='.2f')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
````

## 3. Spearman Correlation
Rank-based, robust to outliers.

````python
# Compare Pearson vs Spearman
pearson = df['Market'].corr(df['Stock_A'])
spearman, pvalue = stats.spearmanr(df['Market'], df['Stock_A'])

print(f'Pearson correlation: {pearson:.4f}')
print(f'Spearman correlation: {spearman:.4f}')
print(f'Spearman p-value: {pvalue:.4f}')
````

## 4. Rolling Correlation

````python
# 60-day rolling correlation
rolling_corr = df['Market'].rolling(60).corr(df['Stock_A'])

plt.figure(figsize=(12, 5))
plt.plot(rolling_corr)
plt.axhline(y=pearson, color='r', linestyle='--', label='Static corr')
plt.title('60-Day Rolling Correlation: Market vs Stock A')
plt.legend()
plt.ylabel('Correlation')
plt.show()
````

corr_matrix = df.corr()
print('Correlation matrix:')
print(corr_matrix.round(3))

# Find minimum correlation (excluding diagonal)
stocks = ['Stock_A', 'Stock_B', 'Stock_C']
min_corr = 1.0
best_pair = None
for i, s1 in enumerate(stocks):
    for s2 in stocks[i+1:]:
        c = abs(corr_matrix.loc[s1, s2])
        if c < min_corr:
            min_corr = c
            best_pair = (s1, s2)

print(f'\nBest diversification: {best_pair}')
print(f'Correlation: {corr_matrix.loc[best_pair[0], best_pair[1]]:.3f}')
print('\nLow correlation = risk reduction when combined!')
