## Learning Objectives
- Create distribution plots with seaborn
- Build categorical comparisons
- Use regression plots
- Create heatmaps and faceted plots

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)

# Generate sample stock data
n = 200
df = pd.DataFrame({
    'Return': np.random.randn(n) * 0.02,
    'Volume': np.random.exponential(1e6, n),
    'Sector': np.random.choice(['Tech', 'Finance', 'Health', 'Energy'], n),
    'MarketCap': np.random.choice(['Large', 'Mid', 'Small'], n)
})
df['Beta'] = np.random.uniform(0.5, 1.5, n)
print('Data ready!')
df.head()
````

## 1. Distribution Plots

````python
# Histogram with KDE
plt.figure(figsize=(10, 5))
sns.histplot(df['Return'], kde=True, bins=30)
plt.title('Return Distribution')
plt.show()
````

````python
# KDE by group
plt.figure(figsize=(10, 5))
for sector in df['Sector'].unique():
    sns.kdeplot(df[df['Sector']==sector]['Return'], label=sector)
plt.legend()
plt.title('Return Distribution by Sector')
plt.show()
````

## 2. Categorical Plots

````python
# Boxplot
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='Sector', y='Return')
plt.title('Return by Sector')
plt.show()
````

````python
# Violin plot
plt.figure(figsize=(10, 5))
sns.violinplot(data=df, x='Sector', y='Return', hue='MarketCap')
plt.title('Return Distribution by Sector and Market Cap')
plt.show()
````

## 3. Regression Plots

````python
# Scatter with regression line
plt.figure(figsize=(10, 6))
sns.regplot(data=df, x='Beta', y='Return', scatter_kws={'alpha':0.5})
plt.title('Beta vs Return with Regression')
plt.show()
````

## 4. Heatmaps

````python
# Correlation heatmap
numeric_cols = df.select_dtypes(include=[np.number])
plt.figure(figsize=(8, 6))
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
````

## 5. Themes and Palettes

````python
# Set theme
sns.set_theme(style='whitegrid')

# Bar plot with custom palette
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='Sector', y='Return', palette='viridis')
plt.title('Mean Return by Sector')
plt.show()
````
