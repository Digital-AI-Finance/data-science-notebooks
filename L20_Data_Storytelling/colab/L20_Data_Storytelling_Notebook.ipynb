````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
np.random.seed(42)

# Sample portfolio data
sectors = ['Tech', 'Finance', 'Health', 'Energy', 'Consumer']
performance = pd.DataFrame({
    'Sector': sectors,
    'Return': [0.25, 0.12, 0.18, -0.05, 0.08],
    'Weight': [0.30, 0.20, 0.20, 0.15, 0.15]
})
print('Performance data ready!')
performance
````

## 1. Before and After
Transform a cluttered chart into a clear message.

````python
# BEFORE: Cluttered
fig, ax = plt.subplots(figsize=(10, 5))
colors = ['red', 'blue', 'green', 'purple', 'orange']
bars = ax.bar(performance['Sector'], performance['Return'], color=colors)
ax.set_ylabel('Return (%)')
ax.set_title('Sector Performance Q4 2024')
ax.grid(True, alpha=0.7)
for bar in bars:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height(),
            f'{bar.get_height():.0%}', ha='center', va='bottom', fontsize=8)
plt.show()
````

````python
# AFTER: Clean with emphasis
fig, ax = plt.subplots(figsize=(10, 5))
colors = ['#0066CC' if r > 0.15 else '#CCCCCC' for r in performance['Return']]
bars = ax.barh(performance['Sector'], performance['Return'], color=colors)
ax.axvline(x=0, color='black', linewidth=0.5)
ax.set_xlabel('Return')
ax.set_title('Tech Leads Sector Performance', fontweight='bold', fontsize=14)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
for bar in bars:
    width = bar.get_width()
    ax.text(width + 0.01, bar.get_y() + bar.get_height()/2,
            f'{width:.0%}', va='center')
plt.tight_layout()
plt.show()
````

## 2. Color for Emphasis

````python
# Highlight the key insight
fig, ax = plt.subplots(figsize=(10, 5))

# Gray for context, blue for focus
colors = ['#0066CC' if s == 'Tech' else '#DDDDDD' for s in performance['Sector']]
bars = ax.bar(performance['Sector'], performance['Return'], color=colors)

# Add annotation
tech_idx = performance[performance['Sector']=='Tech'].index[0]
ax.annotate('Tech outperforms\nall other sectors', 
            xy=(tech_idx, 0.25), xytext=(tech_idx+1.5, 0.22),
            arrowprops=dict(arrowstyle='->', color='black'),
            fontsize=11)

ax.set_ylabel('Return')
ax.set_title('Q4 2024: Tech Dominance', fontweight='bold')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.show()
````

## 3. Annotations

````python
# Time series with event annotations
dates = pd.date_range('2024-01-01', periods=100)
prices = 100 + np.cumsum(np.random.randn(100) * 2)

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(dates, prices, color='#0066CC', linewidth=2)
ax.fill_between(dates, prices, alpha=0.1, color='#0066CC')

# Add event markers
ax.axvline(dates[30], color='red', linestyle='--', alpha=0.5)
ax.annotate('Earnings Miss', xy=(dates[30], prices[30]),
            xytext=(dates[40], prices[30]+10),
            arrowprops=dict(arrowstyle='->', color='red'))

ax.axvline(dates[70], color='green', linestyle='--', alpha=0.5)
ax.annotate('Product Launch', xy=(dates[70], prices[70]),
            xytext=(dates[80], prices[70]+10),
            arrowprops=dict(arrowstyle='->', color='green'))

ax.set_title('Stock Price with Key Events', fontweight='bold')
ax.set_ylabel('Price ($)')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
````
