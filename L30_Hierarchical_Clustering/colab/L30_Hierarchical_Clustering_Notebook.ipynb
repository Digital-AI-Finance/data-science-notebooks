## Learning Objectives
- Build dendrograms
- Compare linkage methods
- Extract cluster labels
- Apply to financial data

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from sklearn.preprocessing import StandardScaler
np.random.seed(42)

# Generate sample stock return data
stocks = ['AAPL', 'MSFT', 'GOOG', 'JPM', 'BAC', 'XOM', 'CVX']
n = 100  # Trading days
returns = np.random.randn(n, len(stocks)) * 0.02
returns[:, :3] += np.random.randn(n, 1) * 0.01  # Tech correlation
returns[:, 3:5] += np.random.randn(n, 1) * 0.01  # Bank correlation

df = pd.DataFrame(returns, columns=stocks)
print('Stock returns ready!')
df.head()
````

## 1. Create Dendrogram

````python
# Compute correlation distance and linkage
corr = df.corr()
dist = 1 - corr  # Correlation distance
Z = linkage(dist, method='ward')

# Plot dendrogram
plt.figure(figsize=(10, 6))
dendrogram(Z, labels=stocks)
plt.title('Stock Clustering Dendrogram')
plt.ylabel('Distance')
plt.show()
````

## 2. Compare Linkage Methods

````python
# Compare different linkage methods
methods = ['single', 'complete', 'average', 'ward']
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

for ax, method in zip(axes.flat, methods):
    Z = linkage(dist, method=method)
    dendrogram(Z, labels=stocks, ax=ax)
    ax.set_title(f'{method.capitalize()} Linkage')
    ax.set_ylabel('Distance')

plt.tight_layout()
plt.show()
````

## 3. Extract Cluster Labels

````python
# Cut dendrogram at K=3
Z = linkage(dist, method='ward')
labels = fcluster(Z, t=3, criterion='maxclust')

result = pd.DataFrame({'Stock': stocks, 'Cluster': labels})
print('Cluster assignments:')
for cluster in sorted(result['Cluster'].unique()):
    members = result[result['Cluster'] == cluster]['Stock'].tolist()
    print(f'  Cluster {cluster}: {members}')
````
