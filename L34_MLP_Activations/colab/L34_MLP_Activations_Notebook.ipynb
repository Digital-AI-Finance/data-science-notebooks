## Learning Objectives
- Build MLPs with scikit-learn
- Understand activation functions
- Compare ReLU, sigmoid, tanh
- Apply to financial data

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPClassifier, MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
np.random.seed(42)

# Generate XOR-like data (not linearly separable)
n = 200
X = np.random.randn(n, 2)
y = ((X[:, 0] > 0) ^ (X[:, 1] > 0)).astype(int)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
print('XOR-like data ready!')
print(f'Class distribution: {np.bincount(y)}')
````

## 1. Visualize Activation Functions

````python
# Plot activation functions
x = np.linspace(-5, 5, 100)

# Define activations
relu = np.maximum(0, x)
sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].plot(x, relu, 'b-', linewidth=2)
axes[0].set_title('ReLU')
axes[0].axhline(0, color='gray', linestyle='--')
axes[0].axvline(0, color='gray', linestyle='--')

axes[1].plot(x, sigmoid, 'g-', linewidth=2)
axes[1].set_title('Sigmoid')
axes[1].axhline(0.5, color='gray', linestyle='--')
axes[1].axvline(0, color='gray', linestyle='--')

axes[2].plot(x, tanh, 'r-', linewidth=2)
axes[2].set_title('Tanh')
axes[2].axhline(0, color='gray', linestyle='--')
axes[2].axvline(0, color='gray', linestyle='--')

plt.tight_layout()
plt.show()
````

## 2. Compare Activations on XOR

````python
# Compare different activation functions
activations = ['relu', 'logistic', 'tanh']
results = {}

for act in activations:
    mlp = MLPClassifier(
        hidden_layer_sizes=(10, 10),
        activation=act,
        max_iter=1000,
        random_state=42
    )
    mlp.fit(X_train, y_train)
    acc = mlp.score(X_test, y_test)
    results[act] = acc
    print(f'{act}: {acc:.3f}')

print(f'\nBest activation: {max(results, key=results.get)}')
````

## 3. Hidden Layer Architecture

````python
# Compare different architectures
architectures = [
    (5,),           # 1 layer, 5 neurons
    (10,),          # 1 layer, 10 neurons
    (10, 5),        # 2 layers
    (10, 10, 5),    # 3 layers
]

for arch in architectures:
    mlp = MLPClassifier(
        hidden_layer_sizes=arch,
        activation='relu',
        max_iter=1000,
        random_state=42
    )
    mlp.fit(X_train, y_train)
    print(f'Architecture {arch}: {mlp.score(X_test, y_test):.3f}')
````
