## Learning Objectives
- Detect overfitting
- Apply regularization (L2)
- Use early stopping
- Compare regularization methods

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split, learning_curve
from sklearn.preprocessing import StandardScaler
np.random.seed(42)

# Generate noisy classification data
n = 300
X = np.random.randn(n, 10)  # 10 features
# Only first 2 features matter
y = ((X[:, 0] + X[:, 1]) > 0).astype(int)

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print('Data ready!')
print(f'Train: {len(X_train)}, Test: {len(X_test)}')
````

## 1. Detect Overfitting

````python
# Train complex model without regularization
mlp_overfit = MLPClassifier(
    hidden_layer_sizes=(100, 50, 25),
    max_iter=500,
    alpha=0.0,  # No regularization
    random_state=42
)
mlp_overfit.fit(X_train_scaled, y_train)

train_acc = mlp_overfit.score(X_train_scaled, y_train)
test_acc = mlp_overfit.score(X_test_scaled, y_test)

print(f'Train accuracy: {train_acc:.3f}')
print(f'Test accuracy:  {test_acc:.3f}')
print(f'Gap: {train_acc - test_acc:.3f}')
if train_acc - test_acc > 0.05:
    print('Warning: Likely overfitting!')
````

## 2. L2 Regularization (Alpha)

````python
# Compare different regularization strengths
alphas = [0.0, 0.001, 0.01, 0.1, 1.0]
results = []

for alpha in alphas:
    mlp = MLPClassifier(
        hidden_layer_sizes=(100, 50, 25),
        max_iter=500,
        alpha=alpha,
        random_state=42
    )
    mlp.fit(X_train_scaled, y_train)
    train_acc = mlp.score(X_train_scaled, y_train)
    test_acc = mlp.score(X_test_scaled, y_test)
    results.append({'alpha': alpha, 'train': train_acc, 'test': test_acc})
    print(f'Alpha={alpha}: Train={train_acc:.3f}, Test={test_acc:.3f}')

# Plot
df_results = pd.DataFrame(results)
plt.figure(figsize=(8, 5))
plt.plot(df_results['alpha'], df_results['train'], 'o-', label='Train')
plt.plot(df_results['alpha'], df_results['test'], 's-', label='Test')
plt.xscale('log')
plt.xlabel('Alpha (L2 regularization)')
plt.ylabel('Accuracy')
plt.legend()
plt.title('Effect of Regularization')
plt.show()
````

## 3. Early Stopping

````python
# Early stopping with validation
mlp_early = MLPClassifier(
    hidden_layer_sizes=(100, 50, 25),
    max_iter=500,
    early_stopping=True,
    validation_fraction=0.1,
    n_iter_no_change=10,
    random_state=42
)
mlp_early.fit(X_train_scaled, y_train)

print(f'Stopped at iteration: {mlp_early.n_iter_}')
print(f'Best validation score: {mlp_early.best_validation_score_:.3f}')
print(f'Test accuracy: {mlp_early.score(X_test_scaled, y_test):.3f}')
````
