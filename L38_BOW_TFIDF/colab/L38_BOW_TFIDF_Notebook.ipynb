## 1. Setup and Imports

````python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import matplotlib.pyplot as plt

print("Libraries loaded successfully!")
````

## 2. Sample Financial Documents

````python
# Sample financial news headlines
documents = [
    "Stock market rises on strong earnings reports",
    "Federal Reserve raises interest rates",
    "Tech stocks fall amid market volatility",
    "Strong earnings boost investor confidence",
    "Market volatility concerns investors",
    "Interest rates impact stock valuations"
]

for i, doc in enumerate(documents):
    print(f"Doc {i+1}: {doc}")
````

## 3. Bag of Words (CountVectorizer)

````python
# Create BOW representation
count_vec = CountVectorizer()
bow_matrix = count_vec.fit_transform(documents)

# Display vocabulary
print("Vocabulary size:", len(count_vec.vocabulary_))
print("\nFeature names:")
print(count_vec.get_feature_names_out())
````

````python
# View BOW matrix as DataFrame
bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=count_vec.get_feature_names_out(),
    index=[f"Doc {i+1}" for i in range(len(documents))]
)
print("BOW Matrix:")
bow_df
````

## 4. TF-IDF Vectorization

````python
# Create TF-IDF representation
tfidf_vec = TfidfVectorizer()
tfidf_matrix = tfidf_vec.fit_transform(documents)

# View TF-IDF matrix
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray().round(3),
    columns=tfidf_vec.get_feature_names_out(),
    index=[f"Doc {i+1}" for i in range(len(documents))]
)
print("TF-IDF Matrix:")
tfidf_df
````

## 5. N-grams

````python
# Bigram vectorizer
bigram_vec = TfidfVectorizer(ngram_range=(1, 2), max_features=20)
bigram_matrix = bigram_vec.fit_transform(documents)

print("Bigram features (sample):")
for feat in bigram_vec.get_feature_names_out():
    print(f"  - {feat}")
````

## 6. Document Similarity

````python
# Compute cosine similarity between documents
similarity_matrix = cosine_similarity(tfidf_matrix)

# Display as heatmap
fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(similarity_matrix, cmap='Blues')
ax.set_xticks(range(len(documents)))
ax.set_yticks(range(len(documents)))
ax.set_xticklabels([f"Doc {i+1}" for i in range(len(documents))])
ax.set_yticklabels([f"Doc {i+1}" for i in range(len(documents))])
plt.colorbar(im, label='Cosine Similarity')
ax.set_title('Document Similarity (TF-IDF)')

# Add text annotations
for i in range(len(documents)):
    for j in range(len(documents)):
        ax.text(j, i, f"{similarity_matrix[i,j]:.2f}", 
                ha='center', va='center', fontsize=9)

plt.tight_layout()
plt.show()
````

## Key Takeaways

1. **BOW** counts word frequencies but ignores word importance
2. **TF-IDF** weights words by importance (rare words get higher weights)
3. **N-grams** capture word sequences ("not good" vs "good")
4. Use `TfidfVectorizer(max_features=N)` to limit vocabulary size
5. TF-IDF is the standard baseline for text classification
