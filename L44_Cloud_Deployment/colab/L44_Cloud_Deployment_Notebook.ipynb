## 1. Requirements File Generation

````python
# Generate requirements.txt content
requirements = """
fastapi==0.104.1
uvicorn==0.24.0
scikit-learn==1.3.2
numpy==1.26.2
joblib==1.3.2
pydantic==2.5.2
""".strip()

print("requirements.txt:")
print(requirements)
````

## 2. Dockerfile Structure

````python
dockerfile = """
# Use Python base image
FROM python:3.10-slim

# Set working directory
WORKDIR /app

# Copy requirements first (for caching)
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# Copy application code
COPY . .

# Expose port
EXPOSE 8000

# Start command
CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
"""

print("Dockerfile:")
print(dockerfile)
````

## 4. GitHub Actions CI/CD

````python
github_actions = """
# .github/workflows/deploy.yml
name: Deploy to Production

on:
  push:
    branches: [main]

jobs:
  test:
    runs-on: ubuntu-latest
    steps:
      - uses: actions/checkout@v3
      - uses: actions/setup-python@v4
        with:
          python-version: '3.10'
      - run: pip install -r requirements.txt
      - run: pytest tests/

  deploy:
    needs: test
    runs-on: ubuntu-latest
    steps:
      - uses: actions/checkout@v3
      - name: Deploy to Heroku
        uses: akhileshns/heroku-deploy@v3.12.14
        with:
          heroku_api_key: ${{ secrets.HEROKU_API_KEY }}
          heroku_app_name: "your-app-name"
          heroku_email: "your@email.com"
"""

print("GitHub Actions workflow:")
print(github_actions)
````

## 6. Streamlit Cloud Deployment

````python
streamlit_app = '''
# app.py for Streamlit Cloud
import streamlit as st
import joblib
import numpy as np

@st.cache_resource
def load_model():
    return joblib.load("model.pkl")

model = load_model()

st.title("ML Prediction App")

# Input features
col1, col2 = st.columns(2)
with col1:
    feature1 = st.slider("Feature 1", -5.0, 5.0, 0.0)
    feature2 = st.slider("Feature 2", -5.0, 5.0, 0.0)
with col2:
    feature3 = st.slider("Feature 3", -5.0, 5.0, 0.0)

# Predict button
if st.button("Predict"):
    features = np.array([[feature1, feature2, feature3]])
    prediction = model.predict(features)[0]
    probability = model.predict_proba(features)[0].max()
    
    st.metric("Prediction", prediction)
    st.metric("Confidence", f"{probability:.2%}")
'''

print("Streamlit Cloud app:")
print(streamlit_app)
print("\nDeploy: Connect GitHub repo to share.streamlit.io")
````
