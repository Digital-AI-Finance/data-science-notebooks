## 2. Reproducibility Essentials

````python
import numpy as np
import random

# Set all random seeds
def set_seed(seed=42):
    """Set random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    # If using TensorFlow:
    # tf.random.set_seed(seed)
    # If using PyTorch:
    # torch.manual_seed(seed)
    print(f"Random seed set to {seed}")

set_seed(42)

# Generate some random numbers to verify
print("Random samples:", np.random.rand(3).round(4))
````
