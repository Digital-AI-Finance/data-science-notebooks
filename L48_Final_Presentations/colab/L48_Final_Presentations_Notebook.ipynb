## 2. Creating Clear Visualizations

````python
import matplotlib.pyplot as plt
import numpy as np

# Good visualization practices
np.random.seed(42)

# Sample model comparison data
models = ['Logistic Reg', 'Random Forest', 'XGBoost', 'Neural Net']
accuracy = [0.72, 0.78, 0.82, 0.80]
colors = ['#ADADE0', '#ADADE0', '#3333B2', '#ADADE0']  # Highlight best

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.bar(models, accuracy, color=colors, edgecolor='#3333B2', linewidth=2)

# Add value labels on bars
for bar, acc in zip(bars, accuracy):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
            f'{acc:.0%}', ha='center', va='bottom', fontsize=14, fontweight='bold')

ax.set_ylabel('Accuracy', fontsize=14)
ax.set_title('Model Comparison: XGBoost Achieves Best Performance', fontsize=16, fontweight='bold')
ax.set_ylim(0, 1)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

print("\nGood chart practices:")
print("  - Clear title with key insight")
print("  - Highlight the important data")
print("  - Large, readable fonts")
print("  - Remove chart junk")
````

## 3. Results Summary Table

````python
import pandas as pd

# Create results summary
results = pd.DataFrame({
    'Model': ['Logistic Regression', 'Random Forest', 'XGBoost', 'Neural Network'],
    'Accuracy': [0.72, 0.78, 0.82, 0.80],
    'Precision': [0.70, 0.76, 0.81, 0.79],
    'Recall': [0.74, 0.80, 0.83, 0.81],
    'F1 Score': [0.72, 0.78, 0.82, 0.80],
    'Training Time (s)': [0.5, 5.2, 12.3, 45.6]
})

# Format for presentation
styled = results.style.format({
    'Accuracy': '{:.0%}',
    'Precision': '{:.0%}',
    'Recall': '{:.0%}',
    'F1 Score': '{:.0%}',
    'Training Time (s)': '{:.1f}s'
}).highlight_max(subset=['Accuracy', 'F1 Score'], color='#ADADE0')

print("Results Summary:")
display(results)
````
